# Ejemplo Distribución binomial.
Ana es una estudiante universitaria que se enfrenta a un examen de selección múltiple que consta de 20 preguntas. Cada pregunta tiene 5 opciones, y solo una de ellas es la respuesta correcta. Ana está un poco nerviosa porque no ha tenido tiempo de estudiar a fondo. Decide responder al azar, eligiendo las opciones de manera aleatoria para cada pregunta. Se plantea:
   
1.	¿Cuál es la probabilidad de que Ana obtenga exactamente 10 respuestas correctas?
2. ¿Cuál es la probabilidad de que Ana acierte al menos 2 respuestas?
3. ¿Cuál es la probabilidad de que Ana tenga entre 2 y 9 respuestas correctas?



n= 20

p 0.2

```python
#a)  Hallemos P(X=10)

P_x_igual_10 = binom.pmf(10, 20, 0.2)
print(P_x_igual_10)
```

```python
#b) Hallemos P(X>=2) = 1- P(X<=1)

P_x_mayor_2 = 1- binom.cdf(1,20,0.2)
print(P_x_mayor_2)
```

# c) Hallmemos $$P(2\leq X\leq 9) $$

```python
P_x_entre_2_y_9 = binom.cdf(9, 20, 0.2) - binom.cdf(1, 20, 0.2)
print(P_x_entre_2_y_9)
```
