```python
# Install dependencies as needed:
# pip install kagglehub[pandas-datasets]
import kagglehub
from kagglehub import KaggleDatasetAdapter

# Set the path to the file you'd like to load
file_path = "ObesityDataSet_raw_and_data_sinthetic.csv"

# Load the latest version
df = kagglehub.dataset_load(
  KaggleDatasetAdapter.PANDAS,
  "fatemehmehrparvar/obesity-levels",
  file_path,
  # Provide any additional arguments like
  # sql_query or pandas_kwargs. See the
  # documenation for more information:
  # https://github.com/Kaggle/kagglehub/blob/main/README.md#kaggledatasetadapterpandas
)

```

```python
df.head()
```

```python
# de participantes( Elementos en el espacio muestral)

N = df.shape[0]
print(N)
```

```python
# Probabilidad de que si se selecciona un participante en la encuesta, esta sea mujer.

#P(Sea Mujer)
```

```python
Mujeres = df[df['Gender'] == 'Female']

P_Mujeres = Mujeres.shape[0] / N
print(P_Mujeres)
```

```python
# Hallemos la probabilidad de que si se selecciona uno de los participantes, este sea un fumador.


P_fumador = (df['SMOKE'] == 'yes').sum() /N # Aca utilizamos la probabilidad clásica de Laplace.
```

```python
print(P_fumador)
```

```python
# Hallemos la probabilidad de que si se selecciona uno de los participants esta sea mujer y fume.

# Ya tenemos filtrada las mujeres en la variable  Mujeres


P_Mujer_Fume = (Mujeres['SMOKE'] == 'yes').sum() / N

print(P_Mujer_Fume)
```

```python
# Hallemos la probabilidad de que si es mujer, sea fumadora.

#Opción 1:

# Restringimos nuestro espacio muestral solo a la mujeres:


mujeres_fumadoras = ((df["Gender"] == "Female") & (df["SMOKE"] == "yes")).sum()
cantidad_mujeres = (df["Gender"] == "Female").sum()
print("Probabilidad de fumar sabiendo que es mujer:", mujeres_fumadoras / cantidad_mujeres)   # Mujeres fumadoras/ Mujeres



```

```python
# Opción 2:

#Ya tenemos algunos resultados

P_Fume_dado_Mujer_2 =    P_Mujer_Fume      / P_Mujeres        # P(Mujer y Fume) / P(Mujer)

print(P_Fume_dado_Mujer_2 )
```
