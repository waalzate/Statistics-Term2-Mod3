```python
# import packages

import matplotlib.pyplot as plt

```

# Ejemplo: Lanzamiento de una moneda tres veces


```python
# Grafiquemos la función de masa de probabilidad

x = [0, 1, 2, 3]

probabilidades = [1/8, 3/8, 3/8, 1/8]

plt.bar(x,probabilidades)
plt.title('Distribución de probabilidad \n Lanzamiento de una moneda tres veces')
plt.xlabel('Número de caras')
plt.ylabel('Probabilidad')
plt.show()
```
